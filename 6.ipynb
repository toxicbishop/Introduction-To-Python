# 🗂️ File Content Sorter

This notebook reads lines from an input file, sorts them alphabetically, and writes the sorted contents to a new output file.

### 💡 Instructions
Ensure an input text file exists. Provide the input and output file names when prompted.


```python
infile = input("Enter input file name: ") 
fh_in = open(infile,"r") 
data=fh_in.readlines() 
lst = []  
for i in range(len(data)): 
    lst.append(data[i].strip()) 
lst.sort() 
fh_in.close() 
outfile=input("Enter output file name: ") 
fh_out = open(outfile, "w") 
for i in lst: 
    fh_out.write(i) 
    fh_out.write("\n") 
fh_out.close() 
```
