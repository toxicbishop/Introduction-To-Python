# 🔢 Digit Frequency Counter

This notebook counts and displays the frequency of each digit in a given multi-digit number.

### 💡 Instructions
Enter a multi-digit number to count the occurrence of each digit.


```python
str1 = input("Enter a multi-digit number: ") 
d = dict() 
for c in str1: 
    if c in d.keys(): 
        d[c] = d[c] + 1 
    else: 
        d[c] = 1 
for k,v in d.items(): 
    print("Number ",k," has frequency: ",v) 
```
